```python
import pandas as pd 
import matplotlib as plt 
```

```python
student_data = 'CSV/Students_exams_cols_cleanup.csv'
```

```python
df = pd.read_csv(student_data)
df.head(20)
```

```python
for i in df:
    print(i)
```

```python
df.columns = [col.lower() for col in df]
df.columns
```

```python
for col in df.columns:
    if df[col].dtype == "object":
        df[col]=df[col].str.replace(" ","_")
df.head()
```

```python
df.rename(columns= {
    'math score'     :'math_score' ,
    'reading score'  :'reading_score',
    'writing score'  :'writing_score',
},inplace=True
)
df.head(4)
```

```python
student_df = pd.DataFrame(columns= ['statics','math','reading','writting'])
```

```python
x = df['math_score'].max()
y = df['reading_score'].max() 
z= df['writing_score'].max()

row= {'statics':'max','math':x,'reading':y,'writting':z}
f1=pd.DataFrame.from_records([row])
student_df= pd.concat([student_df,f1],ignore_index=True)
```

```python
x = df['math_score'].min()
y = df['reading_score'].min() 
z= df['writing_score'].min()

row = row= {'statics':'min','math':x,'reading':y,'writting':z}
f1 =pd.DataFrame.from_records([row]) 
student_df= pd.concat([student_df,f1],ignore_index=True)
student_df
```
